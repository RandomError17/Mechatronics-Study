```python
#Exercise 7
import numpy as np
import matplotlib.pyplot as plt

x = np.array([[9, 7, 5, 3, 1], [8, 6, 4, 2, 0]])
print(x)
print("\nLength:", len(x), "\nRows and Columns:", x.shape, "\nDimensions:", x.ndim)
```

```python
#Exercise 8
import numpy as np
import matplotlib.pyplot as plt

a = np.arange(0, 10, 2)
b = np.linspace(0, 10, 5, endpoint=False)
c = np.ones((2,2))
d = np.zeros((2,2))
e = np.eye(2)
f = np.diag(np.array([2,1]))
g = np.empty(3)
print(a, "\n\n", b, "\n\n", c, "\n\n", d, "\n\n", e, "\n\n", f, "\n\n", g)
```

```python
#Exercise 9
import numpy as np
import matplotlib.pyplot as plt

a = np.arange(1, 6)
b = np.arange(-5, 0)
c = np.arange(2,10,2)
d = np.linspace(0,10,15)
print(a, "\n\n", b, "\n\n", c, "\n\n", d)
```

```python
#Exercise 10
import numpy as np
import matplotlib.pyplot as plt
import random

x = np.linspace(0, 10 * np.pi, 100)
y = np.cos(x)
plt.plot(x, y)
plt.show()

img = np.random.default_rng(27446968).random((50,50))
plt.imshow(img, cmap=plt.cm.gray)
plt.colorbar()


```

```python
#Exercise 11
import numpy as np

x = np.arange(6) + np.arange(0, 51, 10)[:, np.newaxis]
odd_back = x[::-1, ::-2]
even_forward = x[:, ::2]
print(x, "\n\n", odd_back, "\n\n", even_forward)
```

```python
#Exercise 12
import numpy as np

x = np.ones((4,4), dtype=int)
x[3, 1] = 6
x[2, 3] = 2
print(x)

y = np.zeros((6,5))
y[1:] = np.diag(np.arange(2,7))
print("\n", y)
```
