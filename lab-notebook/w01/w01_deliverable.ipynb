***Goal***: Measure 20 identical resistors and see how much their resistance varies. I will use the measurements to find the mean and sample standard deviation and make a histogram. I will then use these results to determine whether the resistors are within their 1% tolerance.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

data = np.array([997, 1003, 995, 1001, 1006, 998, 1004, 992, 999, 1007, 996, 1002, 994, 1000, 1005, 997, 1008, 993, 1001, 999])
mean = np.mean(data)
std = np.std(data, ddof=1)

print("Number of measurements:", len(data))
print(f"Mean resistance: {mean:.2f} Ohms")
print(f"Sample standard deviation: {std:.2f} Ohms")

plt.figure(figsize=(8, 5))
plt.hist(data, bins=8, edgecolor="black")
plt.axvline(mean, linestyle="--", label=f"Mean = {mean:.2f} Ohms")

plt.xlabel("Resistance (Ohms)")
plt.ylabel("Frequency")
plt.title("Resistance Measurements of 20 Nominally Identical Resistors")

plt.legend()
plt.show()
```

***Report***: The resistors meet their stated 1% tolerance. For a 1000 ohm resistor, the range should be 990–1010 ohms. All 20 resistors were within this range, with the lowest reading being 992 ohms and the highest being 1008 ohms. The average was 999.85 ohms, which is very close to the nominal value. Therefore, the resistors meet the stated tolerance.
