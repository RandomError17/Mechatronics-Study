```python
#Exercise 27
import numpy as np
import matplotlib.pyplot as plt

n = 256
X = np.linspace(-np.pi, np.pi, n)
Y = np.sin(2 * X)

plt.plot(X, Y + 1, color='blue', alpha=1.00)
plt.plot(X, Y - 1, color='blue', alpha=1.00)
plt.fill_between(X, 1, Y+1, color='blue', alpha=0.4)
plt.fill_between(X, -1, Y-1, (Y-1) < -1, color='red', alpha=0.4)
plt.fill_between(X, -1, Y-1, (Y-1) > -1, color='blue', alpha=0.4)

```

```python
#Exercise 28
import numpy as np
import matplotlib.pyplot as plt

n = 1024
rng = np.random.default_rng()
X = rng.normal(0,1,n)
Y = rng.normal(0,1,n)
T = np.arctan2(Y, X)

plt.axes((0.05, 0.05, 1, 1))
plt.scatter(X,Y,s=75,c=T,alpha=0.5)

plt.xlim(-1.25, 1.25)
plt.ylim(-1.25,1.25)
```

```python
#Exercise 29
import numpy as np
import matplotlib.pyplot as plt

n = 12
X = np.arange(n)
rng = np.random.default_rng()
Y1 = (1 - X / float(n)) * rng.uniform(0.5, 1.0, n)
Y2 = (1 - X / float(n)) * rng.uniform(0.5, 1.0, n)

plt.bar(X, +Y1, facecolor='#9999ff', edgecolor='white')
plt.bar(X, -Y2, facecolor='#ff9999', edgecolor='white')

for x, y in zip(X, Y1):
    plt.text(x, y, '%.2f' % y, ha='center', va='bottom')

for x, y in zip(X, Y2):
    plt.text(x, -y, '%.2f' % y, ha='center', va='top')

plt.ylim(-1.25, +1.25)
```

```python
#Exercise 30
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return (1 - x / 2 + x ** 5 + y ** 3) * np.exp(-x ** 2 -y ** 2)

n = 256
x = np.linspace(-3, 3, n)
y = np.linspace(-3, 3, n)
X, Y = np.meshgrid(x, y)

plt.contourf(X, Y, f(X, Y), 8, alpha=.75, cmap='hot')
C = plt.contour(X, Y, f(X, Y), 8, colors='black', linewidth=.1)
plt.clabel(C, inline=1)
```

```python
#Exercise 31
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return (1 - x / 2 + x ** 5 + y ** 3) * np.exp(-x ** 2 - y ** 2)

n = 10
x = np.linspace(-3, 3, 4 * n)
y = np.linspace(-3, 3, 3 * n)
X, Y = np.meshgrid(x, y)
plt.imshow(f(X, Y), cmap ='gray')
plt.axes((0.025, 0.025, 0.95, 0.95))
plt.colorbar()
```
